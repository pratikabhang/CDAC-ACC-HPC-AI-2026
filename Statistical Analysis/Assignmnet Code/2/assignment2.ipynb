```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\prati\Downloads\diabetes.csv")

df.head()
```

```python
df.isnull().sum()
```

Median imputation for BP-related columns

```python
df.info()
```

```python
df.columns
```

Median imputation for BP columns

```python
df['bp.1s'] = df['bp.1s'].fillna(df['bp.1s'].median())

df['bp.1d'] = df['bp.1d'].fillna(df['bp.1d'].median())

df['bp.2s'] = df['bp.2s'].fillna(df['bp.2s'].median())

df['bp.2d'] = df['bp.2d'].fillna(df['bp.2d'].median())
```

Mean imputation for all other numerical columns

```python
df['chol'] = df['chol'].fillna(df['chol'].mean())

df['stab.glu'] = df['stab.glu'].fillna(df['stab.glu'].mean())

df['hdl'] = df['hdl'].fillna(df['hdl'].mean())

df['ratio'] = df['ratio'].fillna(df['ratio'].mean())

df['glyhb'] = df['glyhb'].fillna(df['glyhb'].mean())

df['age'] = df['age'].fillna(df['age'].mean())

df['height'] = df['height'].fillna(df['height'].mean())

df['weight'] = df['weight'].fillna(df['weight'].mean())

df['waist'] = df['waist'].fillna(df['waist'].mean())

df['hip'] = df['hip'].fillna(df['hip'].mean())

df['time.ppn'] = df['time.ppn'].fillna(df['time.ppn'].mean())
```

```python
df.isnull().sum()
```

Crosstab – Location with Gender

```python
pd.crosstab(df['location'], df['gender'])
```

Crosstab – Gender with Frame

```python
pd.crosstab(df['gender'], df['frame'])
```

Pie chart – Frame

```python
df['frame'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Frame Distribution')
plt.ylabel('')
plt.show()
```

Pie chart – Gender

```python
df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Gender Distribution')
plt.ylabel('')
plt.show()
```
