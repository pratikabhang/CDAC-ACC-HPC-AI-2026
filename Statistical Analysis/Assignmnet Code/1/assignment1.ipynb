## 1. Mean, Standard Deviation, Median and Quartiles

## LOTSIZE

```python
print("Mean:", df["lotsize"].mean())
```

```python
print("Standard Deviation:", df["lotsize"].std())
```

```python
print("Median:", df["lotsize"].median())
```

```python
print("Q1:", df["lotsize"].quantile(0.25))
print("Q2:", df["lotsize"].quantile(0.50))
print("Q3:", df["lotsize"].quantile(0.75))
```

## PRICE

```python
print("Mean:", df["price"].mean())
```

```python
print("Standard Deviation:", df["price"].std())
```

```python
print("Median:", df["price"].median())
```

```python
print("Q1:", df["price"].quantile(0.25))
print("Q2:", df["price"].quantile(0.50))
print("Q3:", df["price"].quantile(0.75))
```

## 2. Mean, Standard Deviation, Median and Quartiles Grouped by airco

## LOTSIZE

```python
for group, data in df.groupby("airco"):
    print("\nAirco:", group)
    print("Mean:", data["lotsize"].mean())
    print("Standard Deviation:", data["lotsize"].std())
    print("Median:", data["lotsize"].median())
    print("Q1:", data["lotsize"].quantile(0.25))
    print("Q2:", data["lotsize"].quantile(0.50))
    print("Q3:", data["lotsize"].quantile(0.75))
```

## PRICE

```python
for group, data in df.groupby("airco"):
    print("\nAirco:", group)
    print("Mean:", data["price"].mean())
    print("Standard Deviation:", data["price"].std())
    print("Median:", data["price"].median())
    print("Q1:", data["price"].quantile(0.25))
    print("Q2:", data["price"].quantile(0.50))
    print("Q3:", data["price"].quantile(0.75))
```
